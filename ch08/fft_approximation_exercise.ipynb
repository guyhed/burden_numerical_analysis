# FFT Expansion and Frequency Analysis Exercise
Based on Burden & Faires, *Numerical Analysis*, Chapter 8


**Exercise.**  
Consider the function  

\[
f(x)=e^{-x^{2}}+0.5\cos(20\pi x), \qquad x\in[-1,1].
\]

1. Sample \(f\) at \(N=256\) equally‑spaced points on \([-1,1]\).  
2. Use the Fast Fourier Transform (FFT) to compute the discrete Fourier coefficients \(\hat f_k\).  
3. Plot the amplitude spectrum \(|\hat f_k|\) against frequency. Identify the dominant frequencies.  
4. Apply a low‑pass filter by zeroing all coefficients with \(|k|>20\). Reconstruct the signal with the inverse FFT (IFFT).  
5. Plot the original and the filtered signals together and compute the root‑mean‑square error (RMSE) between them.  

*Hint:*  Use `numpy.fft.fftfreq` to obtain the frequency indices that correspond to the FFT output order.



---  
## Solution Outline  

The notebook below walks through the sampling, FFT computation, spectrum analysis, filtering, and reconstruction, following the steps of the exercise. All plots are produced with `matplotlib` (default settings) so that each figure appears as a separate output.


```
import numpy as np
import matplotlib.pyplot as plt
```

```
# Sampling parameters
N = 256
x = np.linspace(-1, 1, N, endpoint=False)

# Define the function f(x)
f = np.exp(-x**2) + 0.5 * np.cos(20 * np.pi * x)

# Inspect first few samples
f[:8]
```

```
# Compute the FFT and corresponding frequencies
F = np.fft.fft(f)
freq = np.fft.fftfreq(N, d=(x[1] - x[0]))  # cycles per unit length
```

```
plt.figure()
plt.plot(x, f, label="Original sample")
plt.title("Sampled function in time domain")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```
plt.figure()
plt.stem(freq, np.abs(F), use_line_collection=True)
plt.title("Amplitude spectrum |F(k)|")
plt.xlabel("Frequency (cycles per unit)")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```
# Low‑pass filter: keep modes with |k| <= 20
F_filtered = F.copy()
mask = np.abs(freq) > 20
F_filtered[mask] = 0

# Inverse FFT for reconstruction
f_filtered = np.fft.ifft(F_filtered).real
```

```
plt.figure()
plt.plot(x, f, label="Original", linewidth=1)
plt.plot(x, f_filtered, label="Filtered (|k|<=20)", linewidth=1)
plt.title("Original vs. filtered signal")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True)
plt.show()
```

```
rmse = np.sqrt(np.mean((f - f_filtered) ** 2))
print(f"RMSE between original and filtered signal: {rmse:.4e}")
```


### Discussion  

The amplitude spectrum shows two distinct frequency regions:  
* A narrow band around zero frequency (the Gaussian component).  
* Peaks at \(\pm 10\) cycles per unit length (because the term is \(\cos(20\pi x) = \cos(2\pi \cdot 10 x)\)).  

By filtering out modes with \(|k|>20\) we effectively preserve all significant energy of the signal, so the reconstruction error is very small (see RMSE). Increasing the cutoff would include more high‑frequency noise, while decreasing it would attenuate the cosine component.
