Chebyshev Polynomial Approximation Exercise Solution

This notebook provides a worked solution to the exercise:

Approximate the function $f(x)=e^{x}$ on the interval $[-1,1]$ by a degree‑8 Chebyshev polynomial.  Compute the Chebyshev coefficients, evaluate the approximation, plot the results and the absolute error, and report the maximum error.

```python
import numpy as np
import matplotlib.pyplot as plt
from numpy.polynomial import chebyshev as cheb
```

```python
# function to approximate
f = lambda x: np.exp(np.sin(np.pi*x))

# degree of approximation
n = 8

# choose many Chebyshev nodes of the first kind for sampling
m = 100  # number of sample points (m > n)
k = np.arange(m)
x_sample = np.cos(np.pi * (k + 0.5) / m)
y_sample = f(x_sample)

# compute Chebyshev coefficients up to degree n
c = cheb.chebfit(x_sample, y_sample, n)
c
```

```python
# evaluation grid
x_plot = np.linspace(-1, 1, 400)
p_plot = cheb.chebval(x_plot, c)

# plot f and polynomial approximation
plt.figure()
plt.plot(x_plot, f(x_plot), label='f(x) = exp(x)')
plt.plot(x_plot, p_plot, linestyle='--', label=f'Chebyshev degree {n}')
plt.legend()
plt.title('Chebyshev approximation of exp(x) on [-1,1]')
plt.show()
```

```python
# absolute error
error = np.abs(f(x_plot) - p_plot)

plt.figure()
plt.plot(x_plot, error)
plt.title('Absolute error |f(x) - p_n(x)|')
plt.xlabel('x')
plt.ylabel('Error')
plt.show()

print('Maximum absolute error:', error.max())
```
