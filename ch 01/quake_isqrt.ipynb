```python
import struct

def fast_inv_sqrt(number):
    """
    Calculates the inverse square root using the Quake III algorithm.
    """
    threehalfs = 1.5
    x2 = number * 0.5
    y = number

    # Evil floating point bit level hacking
    i = struct.unpack('i', struct.pack('f', y))[0]
    i = 0x5f3759df - (i >> 1)
    y = struct.unpack('f', struct.pack('i', i))[0]

    # Newton's method
    y = y * (threehalfs - (x2 * y * y))
    # y = y * (threehalfs - (x2 * y * y))

    return y

# Example usage
number = 4.0
result = fast_inv_sqrt(number)
exact = number**(-0.5)
print(f"The inverse square root of {number} is approximately {result} rel error {abs(result-exact)/exact}")
```
