```python
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
size = 100000

random_array = [2**random.uniform(-100,100) for _ in range(size)]

iter_sum = 0.0
for num in random_array:
    iter_sum += num

sorted_array = sorted(random_array)
sorted_sum = 0.0
for num in sorted_array:
    sorted_sum += num

python_sum = sum(random_array)

abs((python_sum - iter_sum)/python_sum), abs((python_sum- sorted_sum)/python_sum), python_sum
```

```python
phi = (1+np.sqrt(5))/2
ophi = 1-phi

n=100
b = np.empty(n)
b[0]=1
b[1]=ophi
for i in range(2,n):
    b[i] = b[i-1] + b[i-2]

x = np.arange(n)
bp = ophi ** x

```

```python
plt.plot(x,np.log(np.abs(b)),'g', x,np.log(np.abs(bp)), 'r+')
```
