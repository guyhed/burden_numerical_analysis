```python
import random

def random_float_array_sum_and_ordered_sum_iterative(size=10, min_val=0.0, max_val=100.0):
    """
    Creates a random array of floats, sums it using iteration, orders the array, and sums it again using iteration.

    Args:
        size (int): The size of the array.
        min_val (float): The minimum value of the random floats.
        max_val (float): The maximum value of the random floats.

    Returns:
        tuple: A tuple containing the original array, the original sum, the sorted array, and the sorted sum.
    """

    if size <= 0:
        return [], 0.0, [], 0.0

    random_array = [2**random.uniform(min_val, max_val) for _ in range(size)]

    # Calculate original sum using iteration
    original_sum = 0.0
    for num in random_array:
        original_sum += num

    sorted_array = sorted(random_array)

    # Calculate sorted sum using iteration
    sorted_sum = 0.0
    for num in sorted_array:
        sorted_sum += num
    
    python_sum = sum(random_array)

    return python_sum, original_sum, sorted_sum

# Example usage:
python_sum, original_sum, sorted_sum = random_float_array_sum_and_ordered_sum_iterative(10000)

print(f"Sum: {python_sum}  error: {abs((python_sum-original_sum)/python_sum)} oredered error: {abs((python_sum-sorted_sum)/python_sum)}")
print(f"Sum: {python_sum}  abs error: {abs((python_sum-original_sum))}  abs oredered error: {abs((python_sum-sorted_sum))}")

```
